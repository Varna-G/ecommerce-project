# Project 2: E-Commerce Funnel & Conversion Optimization

## Business Problem
An e-commerce marketplace wants to understand where customers drop off
in the purchase journey, which segments drive the most revenue, and
whether product and UX changes meaningfully improve conversion.

## Business Questions
1. What is the conversion rate at each stage of the order funnel?
2. Which product categories and states drive the most revenue?
3. Which customer cohorts have the highest repeat purchase rates?
4. Did our simulated checkout redesign significantly improve conversion?
   (p-value, confidence interval, revenue impact)
5. Which customer segments (RFM) contribute the most lifetime value?

## Funnel Definition
Homepage → Product Browse → Order Placed → Order Approved
→ Order Shipped → Order Delivered

## Dataset
- Source  : Olist Brazilian E-Commerce (Kaggle)
- Orders  : ~100,000 real transactions (2016–2018)
- Tables  : 6 relational CSVs joined on order_id / customer_id
- Target metrics: conversion rate, revenue per stage, repeat rate

## Success Metrics
- Funnel drop-off quantified in both % and $ lost
- A/B test result: statistically significant at p < 0.05
- RFM segments mapped to revenue contribution
- All insights tied to a specific revenue or cost number

## Out of Scope
Real-time tracking, web clickstream data, recommendation engine
